```python
import pandas as pd

# Такой код мне дал ИИ чтобы создать данные
data = {
    'Название': [
        'Мастер и Маргарита',
        'Трудно быть богом',
        'Игры, в которые играют люди',
        'Чистый код',
        'Грокаем алгоритмы',
        'Искусство любить'
    ],
    'Автор': [
        'Михаил Булгаков',
        'Аркадий и Борис Стругацкие',
        'Эрик Берн',
        'Роберт Мартин',
        'Адитья Бхаргава',
        'Эрих Фромм'
    ],
    'Год издания': [1967, 1964, 2016, 2019, 2017, 1956],
    'Цена': [450, 380, 620, 1250, 850, 300]
}
df = pd.DataFrame(data)
df.to_csv('books.csv', index=False, encoding='utf-8')
```
